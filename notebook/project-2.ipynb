```python
print("Quantity:", df["Quantity"].unique())
print("Price Per Unit:", df["Price Per Unit"].unique())
print("Total Spent:", df["Total Spent"].unique())
```

```python
numeric_columns = ["Quantity", "Price Per Unit", "Total Spent"]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df.info()
```

```python

df[["Quantity", "Price Per Unit", "Total Spent"]].notnull().all(axis=1).sum()
```

```python
df["calculated_total"] = df["Quantity"] * df["Price Per Unit"]

(df["calculated_total"] == df["Total Spent"]).sum()
```

```python
condition = (
    df["Quantity"].notnull() &
    df["Price Per Unit"].notnull() &
    df["Total Spent"].isnull()
)

condition.sum()
```

```python
df.loc[condition, ["Transaction ID", "Quantity", "Price Per Unit", "Total Spent"]].head(10)
```

```python
df.loc[condition, "Total Spent"] = (
    df.loc[condition, "Quantity"] * df.loc[condition, "Price Per Unit"]
)
```

```python
df["Total Spent"].isnull().sum()
```

```python
df.loc[
    df["Total Spent"].isnull(),
    ["Transaction ID", "Quantity", "Price Per Unit", "Total Spent"]
]
```

```python
df.groupby("Item")["Price Per Unit"].unique()
```

```python
df.loc[
    df["Total Spent"].isnull(),
    ["Transaction ID", "Item", "Quantity", "Price Per Unit", "Total Spent"]
]
```

```python
item_price = (
    df[df["Item"].notna() & 
       ~df["Item"].isin(["UNKNOWN", "ERROR"])]
    .groupby("Item")["Price Per Unit"]
    .first()
)

item_price
```

```python
df["Price Per Unit"] = df["Price Per Unit"].fillna(
    df["Item"].map(item_price)
)
```

```python
df["Price Per Unit"].isnull().sum()
```

```python
df.loc[
    df["Price Per Unit"].isnull(),
    ["Transaction ID", "Item", "Quantity", "Price Per Unit", "Total Spent"]
]
```

```python
condition_price = (
    df["Price Per Unit"].isnull() &
    df["Quantity"].notnull() &
    df["Total Spent"].notnull()
)

condition_price.sum()
```

```python
df.loc[condition_price, "Total Spent"] / df.loc[condition_price, "Quantity"]
```

```python
df.loc[condition_price, "Price Per Unit"] = (
    df.loc[condition_price, "Total Spent"] /
    df.loc[condition_price, "Quantity"]
)
```

```python
df["Price Per Unit"].isnull().sum()
```

```python
df.loc[
    df["Price Per Unit"].isnull(),
    ["Transaction ID", "Item", "Quantity", "Price Per Unit", "Total Spent"]
]
```

```python
df[["Quantity", "Price Per Unit", "Total Spent"]].isnull().sum()
```

```python
condition_quantity = (
    df["Quantity"].isnull() &
    df["Price Per Unit"].notnull() &
    df["Total Spent"].notnull()
)

condition_quantity.sum()
```

```python
df.loc[condition_quantity, "Total Spent"] / df.loc[condition_quantity, "Price Per Unit"]
```

```python
df.loc[condition_quantity, "Quantity"] = (
    df.loc[condition_quantity, "Total Spent"] /
    df.loc[condition_quantity, "Price Per Unit"]
)
```

```python
df["Quantity"].isnull().sum()
```

```python
df.loc[
    df["Quantity"].isnull(),
    ["Transaction ID", "Item", "Quantity", "Price Per Unit", "Total Spent"]
]
```

```python
import numpy as np
```

```python
df["Item"] = df["Item"].replace(["UNKNOWN", "ERROR"], np.nan)
```

```python
df["Item"].isnull().sum()
```

```python
df["Payment Method"].value_counts(dropna=False)
```

```python
df["Payment Method"] = df["Payment Method"].replace(
    ["UNKNOWN", "ERROR"], np.nan
)
```

```python
df["Payment Method"].isnull().sum()
```

```python
df["Location"].value_counts(dropna=False)
```

```python
df["Location"] = df["Location"].replace(
    ["UNKNOWN", "ERROR"], np.nan
)
```

```python
df["Location"].isnull().sum()
```

```python
df["Transaction Date"].value_counts(dropna=False).head(20)
```

```python
df["Transaction Date"] = df["Transaction Date"].replace(
    ["UNKNOWN", "ERROR"], np.nan
)
```

```python
df["Transaction Date"].isnull().sum()
```

```python
df["Transaction Date"] = pd.to_datetime(
    df["Transaction Date"],
    errors="coerce"
)
```

```python
df["calculated_total"] = df["Quantity"] * df["Price Per Unit"]

df["total_check"] = np.isclose(
    df["Total Spent"],
    df["calculated_total"],
    equal_nan=True
)

df["total_check"].value_counts()
```

```python
df.loc[
    ~df["total_check"],
    [
        "Transaction ID",
        "Item",
        "Quantity",
        "Price Per Unit",
        "Total Spent",
        "calculated_total"
    ]
]
```

```python
condition_total = (
    df["Total Spent"].isnull() &
    df["Quantity"].notnull() &
    df["Price Per Unit"].notnull()
)

condition_total.sum()
```

```python
df.loc[condition_total, "Quantity"] * df.loc[condition_total, "Price Per Unit"]
```

```python
df.loc[condition_total, "Total Spent"] = (
    df.loc[condition_total, "Quantity"] *
    df.loc[condition_total, "Price Per Unit"]
)
```

```python
df["Total Spent"].isnull().sum()
```

```python
df.loc[
    df["Total Spent"].isnull(),
    [
        "Transaction ID",
        "Item",
        "Quantity",
        "Price Per Unit",
        "Total Spent"
    ]
]
```

```python
df.drop(columns=["calculated_total", "total_check"], inplace=True)
```
